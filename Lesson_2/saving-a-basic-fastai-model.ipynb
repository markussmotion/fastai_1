## Saving a Pokemon image identifier to deploy

I decided to grab datasets of digimon and pokeon images from kaggle to build a "mon"-identifier.

```python
from fastai.vision.all import *
```

```python
path = ('/Users/markus.scharf/Documents/GitHub/fastai_1/Lesson_2/imgs')
```

Instead of labeling by first letter capitalization like in the fast.ai lesson 2 example, I went for seperated folders.

Now we can create our `DataLoaders`:

```python
dls = ImageDataLoaders.from_folder('.',
    get_image_files(path),
    valid_pct=0.2,
    seed=42,
    item_tfms=Resize(192))
```

... and train our model, a resnet18 (to keep it small and fast):

```python
learn = vision_learner(dls, resnet18, metrics=error_rate)
learn.fine_tune(3)
```

Now we can export our trained `Learner`. This contains all the information needed to run the model:

```python
learn.export('/Users/markus.scharf/Documents/GitHub/fastai_1/Lesson_2/model.pkl')
```
